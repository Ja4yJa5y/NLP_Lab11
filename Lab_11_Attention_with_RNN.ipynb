## Step 1 — Import libraries

```python
import pandas as pd
import torch
from torch import nn
import matplotlib.pyplot as plt
from google.colab import files

torch.manual_seed(42)
```

## Step 2 — Upload the dataset

```python
df = pd.read_csv("/content/Lab_11_QnA_Dataset.csv")
print("Number of questions:", len(df))
print("Number of answer groups:", df["answer_id"].nunique())
display(df.head())
```

## Step 3 — Preprocessing

1. Lowercase each question and remove punctuation.
2. Give each word a number.
3. Make all questions the same length by adding zeros (**padding**).

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Lowercase
df["question"] = df["question"].str.lower()

# 2. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(df["question"])

# Convert words to numbers and add padding
X = pad_sequences(tokenizer.texts_to_sequences(df["question"]), padding="post")

# Answers
y = df["answer_id"].values

# Convert to PyTorch tensors
X = torch.tensor(X, dtype=torch.long)
y = torch.tensor(y, dtype=torch.long)
```

## Step 4 — Build an RNN with attention

- **Embedding** turns word numbers into vectors.
- **RNN** reads the words in sequence.
- **Attention** gives each word a weight.
- **Output** chooses one answer from the dataset.

```python
class RNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = nn.Embedding(len(tokenizer.word_index) + 1, 16)
        self.rnn = nn.RNN(16, 32)
        self.att = nn.Linear(32, 1)
        self.fc = nn.Linear(32, df["answer_id"].nunique())

    def forward(self, x):
        x, _ = self.rnn(self.emb(x))
        weights = torch.softmax(self.att(x), dim=1)
        return self.fc((x * weights).sum(1)), weights.squeeze(-1)

model = RNN()
```

## Step 5 — Train the model
The model learns which answer belongs to each question.

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

for epoch in range(50):
    loss = nn.functional.cross_entropy(model(X)[0], y) #calculate loss to reduce it
    loss.backward() #backpropagation to update parameters
    optimizer.step() #update parameters

print("Training finished!")
```

## Step 6 — See attention
The bar chart shows how much attention the model gives to each question word.

```python
question = df["question"][0]
words = question.split()

with torch.no_grad():
    prediction, weights = model(X[:1])

plt.bar(words, weights[0, :len(words)].tolist())
plt.title("Attention Weights")
plt.show()
```

## Tasks:


Task 1 — Understand Attention Weights
Easy

Modify Step 6 to visualize the attention weights for three different questions from the dataset instead of only the first question.

Instructions:
- Select three questions from the dataset.
- Generate an attention bar chart for each question.
- Identify the word receiving the highest attention weight in each question.

Question: Do the words with the highest attention weights appear important for understanding the question? Explain your observations.

### **Task 1 Answer**

**Yes, the words with the highest attention weights are highly important for understanding the questions.**

**Observations:**
* **Selective Focus:** The model successfully ignores filler/grammatical words (like *"do"*, *"you"*, *"is"*, *"your"*) and places the highest weights on high-entropy semantic keywords (e.g., **'tired?'** at 45.6%, **'around?'** at 95.8%, and **'designer?'** at 83.1%).
* **Core Intent extraction:** By placing the majority of the weight on these descriptors, the model efficiently extracts the exact noun or state that defines the user's inquiry, allowing for highly accurate downstream classification.

```python
# Task 1 - Select three questions from the dataset and visualize their attention weights
indices = [5, 25, 50]

for idx in indices:
    question = df["question"][idx]
    words = question.split()

    with torch.no_grad():
        prediction, weights = model(X[idx:idx+1])

    # Get the weights corresponding only to the non-padded words of the sentence
    word_weights = weights[0, :len(words)].tolist()

    # Identify the word with the highest attention weight
    max_weight_idx = word_weights.index(max(word_weights))
    best_word = words[max_weight_idx]

    # Plot the results
    plt.figure(figsize=(8, 4))
    plt.bar(words, word_weights, color='skyblue')
    plt.title(f"Attention Weights for Question {idx}\nMost attended word: '{best_word}'")
    plt.ylabel("Attention Weight")
    plt.xticks(rotation=45)
    plt.show()

    print(f"Question {idx}: \"{question}\"")
    print(f"Word with highest attention: '{best_word}' (Weight: {word_weights[max_weight_idx]:.4f})\n" + "-"*50)
```

**Observations & Implications:**
* **Inconsistent Focus:** For *"do you get hurt?"*, the model places its highest weight on the helper verb **'do'** (48.5%). For *"do you get tired?"*, it correctly focuses on the adjective **'tired?'** (45.6%).
* **How Attention Works here:** This suggests that while attention mechanisms can isolate keywords, they are sensitive to word position, sequence structure, and patterns in small training datasets. Without extensive pre-training, the model may associate grammatical templates (like starting with *"do"*) with correct classes rather than consistently focusing on deep semantic equivalents.

```python
# Task 2 - Compare Attention for Similar Questions
# We will select two similar questions. For example:
# Index 0: "do you get hurt?"
# Index 5: "do you get tired?"
# Both questions ask about the physical sensations/vulnerability of the AI using different ending words.

similar_indices = [0, 5]

for idx in similar_indices:
    question = df["question"][idx]
    words = question.split()

    with torch.no_grad():
        prediction, weights = model(X[idx:idx+1])

    # Get attention weights for non-padded words
    word_weights = weights[0, :len(words)].tolist()

    # Plot attention weights
    plt.figure(figsize=(8, 4))
    plt.bar(words, word_weights, color='orange')
    plt.title(f"Attention Weights for Question {idx}: \"{question}\"")
    plt.ylabel("Attention Weight")
    plt.xticks(rotation=45)
    plt.show()

    max_weight_idx = word_weights.index(max(word_weights))
    best_word = words[max_weight_idx]
    print(f"Question: \"{question}\"")
    print(f"Most attended word: '{best_word}' (Weight: {word_weights[max_weight_idx]:.4f})\n" + "-"*50)
```
